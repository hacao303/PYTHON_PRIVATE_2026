## Bài 4 — Đếm học viên đạt
Viết `dem_dat(scores)` nhận một **list điểm** và trả về số điểm `>= 5`.

**Kết quả mẫu:** `[8, 4, 9, 6]` → `3`; `[2, 3, 4]` → `0`.

```python
def dem_dat(scores):
    # TODO: dùng vòng lặp và if để đếm
    count = 0
    for score in scores:
        if score >= 5:
            count += 1
    return count

print(dem_dat([8, 4, 9, 6]))
print(dem_dat([2, 3, 4]))
```

## Bài 5 — Chuẩn hóa họ tên
Viết `chuan_hoa_ten(ten)` để bỏ khoảng trắng thừa và viết hoa chữ cái đầu mỗi từ.

**Kết quả mẫu:**
- `"  nGUYEN   van an  "` → `Nguyen Van An`
- `"   tRAN   THI   MAI "` → `Tran Thi Mai`

*Gợi ý:* `split()`, `join()`, `title()`.

```python
def chuan_hoa_ten(ten):
    # TODO
    # 1. Xóa khoảng trắng thừa ở đầu và cuối
    ten = ten.strip()
    # 2. Tách tên thành các từ
    words = ten.split()
    # 3. Viết hoa chữ cái đầu của mỗi từ, viết thường các chữ cái còn lại
    words = [word.capitalize() for word in words]
    # 4. Nối các từ lại thành một chuỗi, cách nhau bởi một khoảng trắng
    ten_chuan_hoa = " ".join(words)
    return ten_chuan_hoa

print(chuan_hoa_ten("  nGUYEN   van an  "))
print(chuan_hoa_ten("   tRAN   THI   MAI "))
```

## Bài 6 — Báo cáo điểm bằng **3 hàm**
Tự hoàn thiện:
1. `tinh_tb(scores)`: trả về điểm trung bình.
2. `xep_loai(tb)`: `>=8` → `Giỏi`, `>=5` → `Đạt`, còn lại → `Cần cố gắng`.
3. `tao_bao_cao(scores)`: **gọi hai hàm trên** rồi trả về dòng báo cáo.

**Kết quả mẫu:** `[8, 4, 9, 6]` → `Điểm TB: 6.75 | Đạt`.

*Dùng `:.2f` để hiển thị hai chữ số thập phân.*

```python
def tinh_tb(scores):
    # TODO
    if not scores:
        return 0
    return sum(scores) / len(scores)


def xep_loai(tb):
    # TODO
    if tb >= 8:
        return "Gioi"
    elif tb >= 5:
        return "Dat"
    else:
        return "Can co gang"


def tao_bao_cao(scores):
    # TODO: gọi tinh_tb() và xep_loai(); trả về chuỗi báo cáo
    tb = tinh_tb(scores)
    loai = xep_loai(tb)
    return f"Diem trung binh: {tb:.2f}, Xep loai: {loai}"


print(tao_bao_cao([8, 4, 9, 6]))
```
